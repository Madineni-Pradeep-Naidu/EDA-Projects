```python
plt.figure(figsize=(8,5))
top10 = a.head(10)
plt.plot(
    top10['discount_percentage'],
    top10['rating'],
    alpha=1,color= 'pink')

plt.title("Top10 Products Discount_prices VS Rating")
plt.xlabel("Discount_prices")
plt.ylabel("Rating")
```

```python
plt.figure(figsize=(8,5))
top10 = a.head(10)
sns.barplot(x=top10['rating'],y=top10['rating_count'],data=a)
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(y=a['rating'])

plt.title('Boxplot of Ratings')
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.subplot(1,2,1)
sns.boxplot(y=a['rating'])

plt.title('Boxplot of Ratings')
top10 = a['actual_price'].nlargest(10)
plt.figure(figsize=(15,5))
plt.subplot(1,2,2)
sns.barplot(x=top10.values,y=top10.index)

plt.title('Top 10 actual Price')
plt.show()
```
