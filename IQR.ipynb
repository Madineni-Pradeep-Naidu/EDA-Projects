```python
Scores = [45, 50, 52, 47, 49, 51, 53, 48, 50, 52,46, 49, 51, 54, 47, 48, 50, 52, 90, 100]
```

```python
import pandas as pd
ss = pd.Series(Scores)
```

```python
ss
```

```python
q1=ss.quantile(0.25)
```

```python
q1
```

```python
q3 = ss.quantile(0.75)
```

```python
print(q3)
```

```python
iqr = q3-q1
```

```python
print(iqr)
```

```python
lb = q1 - 1.5 * iqr
ub = q3  + 1.5 * iqr
```

```python
print(lb)
print(ub)
```

```python
outlier = ss[(ss< lb) | (ss >ub)]
```

```python
outlier
```

```python
Scores
```

```python
ss
```

```python
ss1 = ss[(ss> lb) & (ss <ub)]
```

```python
ss1
```

```python
from scipy.stats import zscore
```

```python
a =  [45,50,52,55,58,60,62,65,70,93,155,550]
```

```python
pd.Series(a)
```

```python
a = zscore(a)
```

```python
print(a)
```

```python
outlier = a[(a < -3) | (a>3)]
```

```python
outlier
```

```python
clean = a[(a > -3) & (a<3)]
```

```python
clean
```

```python
plt.figure(figsize=(13,6))
plt.subplot(1,2,1)
plt.title("before outlier handling")
sns.violinplot(a,color="red")

plt.subplot(1,2,2)
plt.title("After outlier handling")
sns.violinplot(clean,color="red")
```
