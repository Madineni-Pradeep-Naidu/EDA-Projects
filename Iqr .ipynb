```python
for i in a.columns:
    print(i)
    q1 = a[i].quantile(0.25)
    q3 = a[i].quantile(0.75)
    iqr = q3 - q1
    lb = q1 - 1.5 * iqr
    ub = q3 + 1.5 * iqr
    oulier = a[(a[i] < lb) | (a[i] > ub)]
    print("Q1= ",q1)
    print("Q3= ",q3)
    print("IQR= ",iqr)
    print("Lower bound= ",lb)
    print("Upper bound= ",ub)
    print("Outlier= ",oulier)
```
